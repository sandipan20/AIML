```python
def run_scored(batsman):
  vk=stat[stat['batsman']==batsman]
  return vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(3)
```

```python
run_scored('V Kohli')
```

```python
death_over=stat[stat['over']>15]
```

```python
all_bat=death_over.groupby('batsman')['batsman_runs'].count()
x=all_bat>200
batsman_list=all_bat[x].index.tolist()
```

```python
# strike rate =runs scored / balls played
final=stat['batsman'].isin(batsman_list)
```

```python
final = stat[stat['batsman'].isin(batsman_list)]
runs=final.groupby('batsman')['batsman_runs'].sum()
balls=final.groupby('batsman')['batsman_runs'].count()
sr=(runs/balls)*100
sr.sort_values(ascending=False)
```
