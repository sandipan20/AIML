```python
import requests
```

```python
response = requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=8f874d512b9be86dfccf6cfc6171566f")
```

```python
movie=pd.DataFrame(response.json()['results'])[['id','title','original_title','overview']]
```

```python
movie=pd.DataFrame() # created new empty dataframe
```

```python
for i in range(1, 429):
  response = requests.get(f"https://api.themoviedb.org/3/movie/top_rated?api_key=8f874d512b9be86dfccf6cfc6171566f&page={i}")
  if response.status_code == 200:
    movie_data = pd.DataFrame(response.json()['results'])[['id','title','original_title','overview']]
    movie = pd.concat([movie, movie_data], ignore_index=True)
  else:
    print(f"Failed to fetch page {i}")
```

```python
movie.to_csv('movie.csv')
```
