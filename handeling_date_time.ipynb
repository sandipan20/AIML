```python
import numpy as np
import pandas as pd

```

```python
date = pd.read_csv('orders.csv')
time = pd.read_csv('messages.csv')
```

```python
print(date.head())

time.head()
```

```python
print(date.info())
time.info()
```

```python
# Converting to datetime datatype
date['date'] = pd.to_datetime(date['date'])
date.info()
```

```python
date['date_year'] = date['date'].dt.year

date.sample(5)
```

```python
date['date_month_no'] = date['date'].dt.month

date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()

date.head()
```

```python
date['date_day'] = date['date'].dt.day

date.head()
```

```python
# day of week
date['date_dow'] = date['date'].dt.dayofweek

date.head()
```

```python
# day of week - name

date['date_dow_name'] = date['date'].dt.day_name()

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
	
# is weekend?

date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
date['date_week'] = date['date'].dt.isocalendar().week

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# Extract week number
date['date_week'] = date['date'].dt.isocalendar().week.astype(int)

# Display without selected columns
date.drop(
    columns=['product_id', 'city_id', 'orders']
).head()
```

```python
date['date'] = pd.to_datetime(date['date'])

date['quarter'] = date['date'].dt.quarter

date['semester'] = np.where(
    date['quarter'].isin([1, 2]),
    1,
    2
)

date.drop(
    columns=['product_id', 'city_id', 'orders']
).head()
```

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# Months passed
np.round((today - date['date']).dt.days / 30, 0)
```

```python
print(time.info())
# Converting to datetime datatype
time['date'] = pd.to_datetime(time['date'])
time.info()
```

```python
time['hour'] = time['date'].dt.hour
time['min'] = time['date'].dt.minute
time['sec'] = time['date'].dt.second

time.head()
```

```python
time['time'] = time['date'].dt.time

time.head()
```

```python
today - time['date']
```

```python
# in seconds

(today - time['date'])/np.timedelta64(1,'s')
```

```python
# in minutes

(today - time['date'])/np.timedelta64(1,'m')
```

```python
# in hours

(today - time['date'])/np.timedelta64(1,'h')
```
